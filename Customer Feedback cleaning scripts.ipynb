```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore') #ignoring warnings 

df=pd.read_csv('CustomerFeedbackuncleaneddata.csv')
print("Dataset\n")
display(df) #Displaying the Dataframe in table format
```

```python
#checking dataframe shape
print("Number of Rows and Columns in Dataset")
print(df.shape)
```

```python
#checking datatype of dataframe
print("Data types of each columns")
print(df.dtypes)
```

```python
#Rename Feedback_ID,Customer_ID,Review_Date,Product_Category,Sub_Category,Review_Title,Review_Text,Response_Time_Days,Agent_Name
print(df.columns.tolist())#List of Columns
df.columns = df.columns.str.strip()
df.rename(columns={
    'Feedback_ID':'FeedbackID',
    'Customer_ID':'CustomerID',
    'Review_Date':'FeedbackDate',
    'Product_Category':'ProductCategory',
    'Sub_Category':'SubCategory',
    'Review_Title':'FeedbackTitle',
    'Review_Text':'Feedback',
    'Response_Time_Days':'ResponseDays',
    'Agent_Name':'Agent'}, inplace=True)
print("Successfully Renamed Columns")
```

```python
#Remove leading/ trailing spaces
'''df["ProductCategory"] = df["ProductCategory"].str.strip()
df["SubCategory"] = df["SubCategory"].str.strip()
df["Channel"] = df["Channel"].str.strip()
df["Region"] = df["Region"].str.strip()
df["FeedbackTitle"] = df["FeedbackTitle"].str.strip()
df["Feedback"] = df["Feedback"].str.strip()
df["Sentiment"] = df["Sentiment"].str.strip()
df["Agent"] = df["SupportAgent"].str.strip()
df["Priority"] = df["Priority"].str.strip()'''
for column in df.select_dtypes(include="object").columns:
    df[column]=df[column].str.strip()
print("Leading/Trailing spaces removed")
```

```python
#Checking and Removing duplicate rows
print("Number of Duplicate rows in Feedback Dataset\n",df.duplicated().sum()) #number of Duplicated rows 
print("Below is the list of Duplicate rows")
print(display(df[df.duplicated()])) #List of duplicated rows
df.drop_duplicates(inplace=True) #Dropping duplicated rows
print("Successfully removed Duplicated rows")
```

```python
#checking and handling missing value
print("Missing values")
print(df.isnull().sum()) #check missing values
df["FeedbackTitle"].fillna("No Title", inplace=True) #handling missing value in Feedback Title
df["Feedback"].fillna("No Feedback", inplace=True) #handling missing value in Feedback
print("Successfully Value Filled")
```

```python
#checking Blank values
for column in df.select_dtypes(include="object").columns:
    blank_count = (df[column].fillna("").str.strip() == "").sum()
    print(column, ":", blank_count)
```

```python
#Standardize Inconsistent Categorical Values
df["Resolved"] = df["Resolved"].str.lower()
df["Resolved"] = df["Resolved"].replace({
    "y": "yes",
    "n": "no"})
print("Successfully Replaced")
```

```python
#Check unique value of Sentiment
df["Sentiment"] = df["Sentiment"].str.lower()
print("Sentiment values:")
print(df["Sentiment"].value_counts(dropna=False))
```

```python
#df["FeedbackDate"] = pd.to_datetime(df["FeedbackDate"])
df["FeedbackDate"] = pd.to_datetime(
    df["FeedbackDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce") #Changing and handling Date column
print(df["FeedbackDate"].dtype) 
missing_dates = df["FeedbackDate"].isna().sum()
print("Missing/Invalid dates:", missing_dates)
print("Percentage:", round((missing_dates / len(df)) * 100, 2), "%")
```

```python
df.dropna(subset=["FeedbackDate"], inplace=True)
print("Rows after removing missing dates:", len(df))
print("Missing dates after cleaning:", df["FeedbackDate"].isna().sum())
```

```python
#Checking Data range of Rating and Response Days
print("Rating Range")
print("Minimum:", df["Rating"].min())
print("Maximum:", df["Rating"].max())
print("ResponseDays Range")
print("Minimum:", df["ResponseDays"].min())
print("Maximum:", df["ResponseDays"].max())
```

```python
#Validate rating and Response Days
print(df["Rating"].unique())
invalid_ratings = df[(df["Rating"] < 1) | (df["Rating"] > 5)] #checking values below 0 or above 5
print("Invalid Ratings:", len(invalid_ratings))
invalid_response_days = df[df["ResponseDays"] < 0]
print("Invalid ResponseDays:", len(invalid_response_days))
```

```python
#Remove special characters and punctuation
import string 

def remove_punctuation(text):
    if isinstance(text, str):
        for symbol in string.punctuation:
            text=text.replace(symbol, "")
    return text
df["Feedback"]=df["Feedback"].apply(remove_punctuation)
print("Successfully Removed Special Characters")
```

```python
#Standardize letter case
df["ProductCategory"]=df["ProductCategory"].str.lower()
df["SubCategory"]=df["SubCategory"].str.lower()
df["Channel"]=df["Channel"].str.lower()
df["Region"]=df["Region"].str.lower()
df["FeedbackTitle"]=df["FeedbackTitle"].str.title()
df["Agent"]=df["Agent"].str.title()
df["Priority"] = df["Priority"].str.lower()
print("Successfully Changed")
```

```python
#Remove unwanted columns
df.drop(["FeedbackID"],axis=1,inplace=True,errors="ignore")
print("Sucessfully removed unwanted columns")
```

```python
print("Final Rows and Columns:", df.shape)
print("Final Duplicate Rows:", df.duplicated().sum())
print("\nFinal Missing Values:",df.isnull().sum())
print("\nFinal Data Types:\n",df.dtypes)
print("\nFinal Columns:\n",df.columns.tolist())
```

```python
#exporting the csv file
df.to_csv('customerfeedbackcleaneddata.csv', index=False)
print("File Saved")
```
